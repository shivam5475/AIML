```python
# Import required libraries
from transformers import AutoModel
import numpy as np
import soundfile as sf
from IPython.display import Audio

# Load the IndicF5 model from Hugging Face
print("Loading IndicF5 model...")
repo_id = "ai4bharat/IndicF5"
model = AutoModel.from_pretrained(repo_id, trust_remote_code=True)
print("Model loaded successfully.")

# Define inputs for speech generation
text_to_synthesize = "नमस्ते! संगीत की तरह जीवन भी खूबसूरत होता है, बस इसे सही ताल में जीना आना चाहिए."  # Hindi text
ref_audio_path = "/content/MAR_F_HAPPY_00001.wav"  # Reference audio
ref_text = "ਭਹੰਪੀ ਵਿੱਚ ਸਮਾਰਕਾਂ ਦੇ ਭਵਨ ਨਿਰਮਾਣ ਕਲਾ ਦੇ ਵੇਰਵੇ ਗੁੰਝਲਦਾਰ ਅਤੇ ਹੈਰਾਨ ਕਰਨ ਵਾਲੇ ਹਨ, ਜੋ ਮੈਨੂੰ ਖੁਸ਼ ਕਰਦੇ ਹਨ।"  # Punjabi transcript

# Generate speech
print("Generating speech...")
audio = model(
    text_to_synthesize,
    ref_audio_path=ref_audio_path,
    ref_text=ref_text
)
print("Speech generated.")

# Normalize audio if needed and save
print("Saving audio...")
if audio.dtype == np.int16:
    audio = audio.astype(np.float32) / 32768.0  # Convert int16 to float32
sf.write("samples/namaste.wav", np.array(audio, dtype=np.float32), samplerate=24000)
print("Audio saved to samples/namaste.wav")

# Play the audio in Colab
print("Playing audio...")
display(Audio("samples/namaste.wav"))
```
